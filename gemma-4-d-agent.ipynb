```python
import shutil, zipfile
from pathlib import Path

SRC = Path('/kaggle/input/datasets/enginamullah/submission')
OUT = Path('/kaggle/working/submission')

# Zip the folder so agent.yaml sits at the top of the archive
zip_path = shutil.make_archive(str(OUT), 'zip', root_dir=SRC)

# Check the contents
with zipfile.ZipFile(zip_path) as zf:
    names = zf.namelist()
print(zip_path)
print(names)
assert 'agent.yaml' in names, 'agent.yaml is not at the root!'
print('OK: ready to submit')
```
